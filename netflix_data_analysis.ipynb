**Step 1: import libraries and load data**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/netflix_titles.csv')
df
```

```python
df.head()
```

```python
df.tail()
```

**Step 2: Data Overview and missing value**

```python
df.info()
```

```python
df.describe(include="all")
```

```python
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

**Step 3:Handling missing values**

```python
df.isnull().sum()
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage.sort_values(ascending=False)
```

**Step 4:Removing duplicates value**

```python
print("Duplicate Records:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print("Updated Dataset Shape:", df.shape)
```

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Not Rated")
df["duration"] = df["duration"].fillna("Not Available")
df["date_added"] = df["date_added"].fillna("Not Available")
df["listed_in"] = df["listed_in"].fillna("Unknown")
df["description"] = df["description"].fillna("No Description")

print("Missing Values Handled!")
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"].str.strip(),
    format="%B %d, %Y",
    errors="coerce"
)

df["year_added"] = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month

df.head()
```

**Step 6: Exploratory Data Analysis (EDA)**

```python
## Analysis 1 – Movies vs TV Shows

df["type"].value_counts()
```

```python
#visualization

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="type")

plt.title("Movies vs TV Shows on Netflix", fontsize=15)
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.savefig("movies_vs_tvshows.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 2 – Top 10 Countries

countries = (
    df["country"]
    .str.split(",")
    .explode()
    .str.strip()
)

top_countries = (
    countries[countries != "Unknown"]
    .value_counts()
    .head(10)
)

top_countries
```

```python
## Country Visualization

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title("Top 10 Countries by Netflix Titles")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.savefig("top_10_countries.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 3 – Top 10 Genres

genres = (
    df["listed_in"]
    .str.split(",")
    .explode()
    .str.strip()
)

top_genres = genres.value_counts().head(10)

top_genres
```

```python
## genre visualization

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.savefig("top_10_genres.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 4 – Content Ratings

df["rating"].value_counts().head(10)
```

```python
## Rating Visualization

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="rating",
    order=df["rating"].value_counts().head(10).index
)

plt.title("Top 10 Netflix Content Ratings")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.savefig("content_ratings.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 5 – Release Year Trend

yearly_content = df["release_year"].value_counts().sort_index()

yearly_content.tail(20)
```

```python
## Visualization

plt.figure(figsize=(12, 5))

plt.plot(
    yearly_content.index,
    yearly_content.values,
    marker="o"
)

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.savefig("release_year_trend.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 6 – Content Added to Netflix by Year

content_added = df["year_added"].value_counts().sort_index()

content_added
```

```python
## Visualization

plt.figure(figsize=(10, 5))

plt.plot(
    content_added.index,
    content_added.values,
    marker="o"
)

plt.title("Netflix Content Added by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.savefig("content_added_over_time.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
## Analysis 7 – Movie Duration Analysis

movies = df[df["type"] == "Movie"].copy()

movies["duration_minutes"] = pd.to_numeric(
    movies["duration"].str.extract(r"(\d+)")[0],
    errors="coerce"
)

print("Average Duration:",
      round(movies["duration_minutes"].mean(), 2), "minutes")

print("Maximum Duration:",
      movies["duration_minutes"].max(), "minutes")
```

```python
## visualization

plt.figure(figsize=(10, 5))

sns.histplot(
    movies["duration_minutes"].dropna(),
    bins=30,
    kde=True
)

plt.title("Netflix Movie Duration Distribution")
plt.xlabel("Duration in Minutes")
plt.ylabel("Number of Movies")

plt.savefig("movie_duration.png", dpi=300, bbox_inches="tight")
plt.show()
```

**Final Summary Of Analysis**
